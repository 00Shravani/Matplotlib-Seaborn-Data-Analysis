# matplotlib

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# plot(): it is used to create  a basic 2D plot
# plt.plot(x,y)
x = np.arange(1,6)
y = x**2
z = x**3
# marker='o d * . < >'
plt.plot(x,y,color='green',label='sqr',marker='o',markersize=8,markerfacecolor='orange',markeredgecolor='orange',linestyle='solid',linewidth=3)
for i,j in zip(x,y):
  plt.annotate(f'({i},{j})',(i,j),ha='left')

plt.plot(x,z,color='red',label='cube',marker='o',markersize=9,markerfacecolor='orange',markeredgecolor='orange',linestyle='dashed',linewidth=3)
for i,j in zip(x,z):
  plt.annotate(f'({i},{j})',(i,j))

plt.xlabel('val from 1 to 5',color='#8e44ad',fontsize=10)
plt.ylabel('squares from 1 to 5',color='#8e44ad',fontsize=10)
plt.title('SQUARE',color='red',fontsize=15)
plt.legend(shadow=True,facecolor='lightblue',loc='best')
plt.show()
```

```python
a = sns.countplot(x='type',data=df,hue='type',palette=['red','blue'],width=0.1)

#var.containers:   collection of containers
def data_label(a):
  for i in a.containers:
    a.bar_label(i,color='white',fontsize=10,label_type='center',rotation=90) # it traverse through a single container and it represents a datalabel

data_label(a)
```

```python
b = sns.countplot(x='type',data=df,hue='quality',palette='viridis')
plt.title("count vs type by quality")
plt.xlabel('Type')
plt.ylabel('Count')
data_label(b)
plt.show()
```

syntax: sns.boxplot(x='colname',y='colname',data=df,hue='cat_cal')

```python
df.describe()
```

```python
sns.boxplot(x='type',y='alcohol',data=df)
```

```python
#Handling outliers
q1 = df['alcohol'].quantile(0.25)
q3 = df['alcohol'].quantile(0.75)

IQR = q3-q1
lower_bound = q1-1.5*IQR
upper_bound = q3+1.5*IQR

#identifying outliers
outliers = df[(df['alcohol']<lower_bound) | (df['alcohol']>upper_bound)]
outliers


#removing outliers

non_outliers = df[(df['alcohol']>=lower_bound) & (df['alcohol']<=upper_bound)]
non_outliers

#winsorization or capping

out = []
for i in df['alcohol']:
  if i>upper_bound:
    out.append(upper_bound)
  elif i<lower_bound:
    out.append(lower_bound)
  else:
    out.append(i)
df['alcohol'] = out
```

```python
sns.boxplot(x='type',y='alcohol',data=df,hue='type')
```

```python
sns.boxplot(x='type',y='pH',data=df,hue='type')
```

```python
#Handling outliers
q1 = df['pH'].quantile(0.25)
q3 = df['pH'].quantile(0.75)

IQR = q3-q1
lower_bound = q1-0.9*IQR
upper_bound = q3+0.9*IQR

#identifying outliers
outliers = df[(df['pH']<lower_bound) | (df['pH']>upper_bound)]
outliers


#removing outliers

non_outliers = df[(df['pH']>=lower_bound) & (df['pH']<=upper_bound)]
non_outliers

#winsorization or capping

out = []
for i in df['pH']:
  if i>upper_bound:
    out.append(upper_bound)
  elif i<lower_bound:
    out.append(lower_bound)
  else:
    out.append(i)
df['pH'] = out
```

```python
sns.boxplot(x='type',y='pH',data=df,hue='type')
```
